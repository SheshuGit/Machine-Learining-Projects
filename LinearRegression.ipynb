```python
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('Study_vs_Score_data.csv')
df.head(5)
```

```python
x=df.drop('Final_Marks',axis=1)
y=df['Final_Marks']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
print(model.predict([[5]]))
```

```python
from sklearn.metrics import mean_squared_error

error=mean_squared_error(y_pred,y_test)
print(error)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

# Sample dataset (non-linear)
np.random.seed(42)
X = np.sort(5 * np.random.rand(80, 1), axis=0)
y = np.sin(X).ravel() + np.random.randn(80) * 0.1

# Define models
models = {
    "Linear": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.1),
    "Polynomial (Degree 3)": make_pipeline(PolynomialFeatures(3), LinearRegression())
}

# Plot
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='black', label='Data')

X_test = np.linspace(0, 5, 100).reshape(-1, 1)

for name, model in models.items():
    model.fit(X, y)
    y_pred = model.predict(X_test)
    plt.plot(X_test, y_pred, label=name)

plt.title("Linear Regression Algorithms Comparison")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
